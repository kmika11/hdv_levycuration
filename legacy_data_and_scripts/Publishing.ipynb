```python
import os
import requests
import json
from time import sleep
```

```python
base_url = "https://dataverse.harvard.edu"
api_token = "899c1144-ae22-4e5e-b89d-c73e2993694f"
MAJOR_OR_MINOR = "major"
headers = {'X-Dataverse-key': api_token,}
```

#### Find draft datasets 

```python
alias = "levy_photos_1986"

get_cotents_url = '%s/api/dataverses/%s/contents' % (base_url, alias)
r = requests.get(get_cotents_url, headers = headers)
contents = r.json()
```

```python
dois = []
for c in contents['data']:
    dois.append(c['identifier'])
```

```python
def prepend(l, s):
      
    # Using format()
    s += '{0}'
    l = [s.format(i) for i in l]
    return(l)
```

```python
s = "doi:10.7910/"
full_dois = prepend(dois, s)
full_dois
```

```python
#get versions of all datasets
versions = []
for doi in full_dois:
    get_versions_url = "%s/api/datasets/:persistentId/versions?persistentId=%s" % (base_url, doi)
    r = requests.get(get_versions_url, headers = headers)
    versions.append(r.json())
```

```python
versions[2]['data'][0]
```

```python
drafts = []
for i in versions:
    if i['data'][0]['versionState'] == "DRAFT":
        drafts.append(i['data'][0]['datasetPersistentId'])
    else:
        pass
```

```python
len(drafts)
```

```python
drafts
```

### publish draft datasets

```python
published_dois = []

for doi in drafts:
    url_request = "%s/api/datasets/:persistentId/actions/:publish?persistentId=%s&type=%s" % (base_url, doi, MAJOR_OR_MINOR)
    r = requests.post(url_request, headers=headers)
    published_dois.append(json.loads(r.text))
    sleep(20)
```
